#### **Gradient Boosting**

```python
X = df[
    [
        "campaign",
        "pdays"
    ]
]

Y = df[["previous"]]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gbr = GradientBoostingRegressor(
    n_estimators=100,      # number of trees
    learning_rate=0.1,     # contribution of each tree
    max_depth=3,           # complexity of each tree
    random_state=42
)
```

```python
gbr.fit(X_train, Y_train.values.ravel())
```

```python
Y_pred = gbr.predict(X_test)

Y_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(
    Y_test,
    Y_pred
)

mse = mean_squared_error(
    Y_test,
    Y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    Y_test,
    Y_pred
)
```

```python
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R2 score:", round(r2, 2))
```

```python
Y_train_pred = gbr.predict(X_train)

training_r2 = r2_score(
    Y_train,
    Y_train_pred
)

print(
    "Training R2 Score:",
    round(training_r2, 4)
)
```

#### **Overall comarision**

```python
# Overall Model Comparison

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score

# Target variable
Y = df[["previous"]]


# 1. Simple Linear Regression
X_simple = df[["campaign"]]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_simple,
    Y,
    test_size=0.2,
    random_state=42
)

simple_model = LinearRegression()
simple_model.fit(X_train_s, y_train_s)

simple_train_r2 = simple_model.score(X_train_s, y_train_s)
simple_test_r2 = simple_model.score(X_test_s, y_test_s)


# 2. Multiple Linear Regression
X_multiple = df[["campaign", "pdays"]]

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X_multiple,
    Y,
    test_size=0.2,
    random_state=42
)

multiple_model = LinearRegression()
multiple_model.fit(X_train_m, y_train_m)

multiple_train_r2 = multiple_model.score(X_train_m, y_train_m)
multiple_test_r2 = multiple_model.score(X_test_m, y_test_m)


# 3. Gradient Boosting Regression
X_train_g, X_test_g, y_train_g, y_test_g = train_test_split(
    X_multiple,
    Y,
    test_size=0.2,
    random_state=42
)

gbr = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

gbr.fit(X_train_g, y_train_g.values.ravel())

gbr_train_pred = gbr.predict(X_train_g)
gbr_test_pred = gbr.predict(X_test_g)

gbr_train_r2 = r2_score(y_train_g, gbr_train_pred)
gbr_test_r2 = r2_score(y_test_g, gbr_test_pred)


# Model comparison
models = [
    "Simple Linear Regression",
    "Multiple Linear Regression",
    "Gradient Boosting Regression"
]

training_r2 = [
    simple_train_r2,
    multiple_train_r2,
    gbr_train_r2
]

test_r2 = [
    simple_test_r2,
    multiple_test_r2,
    gbr_test_r2
]

comparison = pd.DataFrame({
    "Model": models,
    "Training R²": training_r2,
    "Test R²": test_r2
})

comparison
```

```python
comparison.plot(
    x="Model",
    y=["Training R²", "Test R²"],
    kind="bar",
    figsize=(10, 6)
)

plt.title("Overall Model R² Score Comparison")
plt.xlabel("Regression Models")
plt.ylabel("R² Score")
plt.xticks(rotation=15)
plt.legend()
plt.tight_layout()
plt.show()
```
